```python
import tensorflow as tf
from tensorflow.keras import layers, models

# --- 1. Load and Preprocess Data ---
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.cifar10.load_data()

# Normalize pixel values to be between 0 and 1
X_train = X_train.astype('float32') / 255.0
X_test = X_test.astype('float32') / 255.0

# --- 2. Build AlexNet ---
# We modify the strides and pooling to accommodate 32x32 input
model = models.Sequential([
    # Layer 1: Convolutional
    layers.Conv2D(96, (3, 3), strides=(1, 1), activation='relu', padding='same', input_shape=(32, 32, 3)),
    layers.BatchNormalization(),
    layers.MaxPooling2D(pool_size=(2, 2), strides=(2, 2)),

    # Layer 2: Convolutional
    layers.Conv2D(256, (5, 5), activation='relu', padding='same'),
    layers.BatchNormalization(),
    layers.MaxPooling2D(pool_size=(2, 2), strides=(2, 2)),

    # Layer 3: Convolutional
    layers.Conv2D(384, (3, 3), activation='relu', padding='same'),

    # Layer 4: Convolutional
    layers.Conv2D(384, (3, 3), activation='relu', padding='same'),

    # Layer 5: Convolutional
    layers.Conv2D(256, (3, 3), activation='relu', padding='same'),
    layers.MaxPooling2D(pool_size=(2, 2), strides=(2, 2)),

    # Layer 6: Flatten and Fully Connected
    layers.Flatten(),
    layers.Dense(4096, activation='relu'),
    layers.Dropout(0.5),

    # Layer 7: Fully Connected
    layers.Dense(4096, activation='relu'),
    layers.Dropout(0.5),

    # Layer 8: Output (10 classes for CIFAR-10)
    layers.Dense(10, activation='softmax')
])

# --- 3. Compile and Train ---
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# AlexNet is heavy! Using a smaller batch size if your memory is limited
model.fit(X_train, y_train, epochs=10, batch_size=128, validation_split=0.1)

# --- 4. Evaluate ---
test_loss, test_acc = model.evaluate(X_test, y_test)
print(f"\nAlexNet CIFAR-10 Test Accuracy: {test_acc:.4f}")
```
