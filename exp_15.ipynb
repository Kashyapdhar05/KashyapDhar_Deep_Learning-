```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

# Load your dataset (e.g., 'AAPL.csv')
df = pd.read_csv('Stock_Price_Data.csv')
data = df.filter(['Close']).values

# Scale data between 0 and 1
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_data = scaler.fit_transform(data)

# Create sequences (Windowing)
prediction_days = 60
x_train, y_train = [], []

for x in range(prediction_days, len(scaled_data)):
    x_train.append(scaled_data[x-prediction_days:x, 0])
    y_train.append(scaled_data[x, 0])

x_train, y_train = np.array(x_train), np.array(y_train)
x_train = np.reshape(x_train, (x_train.shape[0], x_train.shape[1], 1))

# Build Model
model = Sequential([
    LSTM(50, return_sequences=True, input_shape=(x_train.shape[1], 1)),
    Dropout(0.2), # Prevents overfitting to noise
    LSTM(50, return_sequences=False),
    Dropout(0.2),
    Dense(25),
    Dense(1)
])

model.compile(optimizer='adam', loss='mean_squared_error')
model.fit(x_train, y_train, batch_size=32, epochs=10)


from tensorflow.keras.preprocessing.sequence import TimeseriesGenerator

# Assume df has 'DATE' and 'MILES'
df = pd.read_csv('Miles_Travelled.csv', index_col='DATE', parse_dates=True)
train_data = df.iloc[:len(df)-12] # Save last 12 months for testing

scaler = MinMaxScaler()
scaled_train = scaler.fit_transform(train_data)

# Window size of 12 to capture the yearly seasonality
n_input = 12
n_features = 1
generator = TimeseriesGenerator(scaled_train, scaled_train, length=n_input, batch_size=1)

# Simple RNN Model
model = Sequential([
    LSTM(150, activation='relu', input_shape=(n_input, n_features)),
    Dense(1)
])

model.compile(optimizer='adam', loss='mse')
model.fit(generator, epochs=30)
```
