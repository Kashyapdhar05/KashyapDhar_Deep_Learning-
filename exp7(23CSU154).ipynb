```python
import tensorflow as tf
from tensorflow.keras import layers, models

# --- 1. Load and Preprocess Data ---
mnist = tf.keras.datasets.mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Fix: Convert to float32 first to allow division/scaling
# Also adding the channel dimension (Samples, Height, Width, 1)
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Reshape to (Samples, 28, 28, 1)
X_train = X_train[..., tf.newaxis]
X_test = X_test[..., tf.newaxis]

# Pad from 28x28 to 32x32 to match LeNet-5 architecture
X_train = tf.pad(X_train, [[0, 0], [2, 2], [2, 2], [0, 0]])
X_test = tf.pad(X_test, [[0, 0], [2, 2], [2, 2], [0, 0]])

# --- 2. Build LeNet-5 ---
model = models.Sequential([
    # C1: Convolutional Layer (6 filters, 5x5 kernel)
    layers.Conv2D(6, (5, 5), activation='relu', input_shape=(32, 32, 1)),

    # S2: Average Pooling Layer (2x2)
    layers.AveragePooling2D((2, 2)),

    # C3: Convolutional Layer (16 filters, 5x5 kernel)
    layers.Conv2D(16, (5, 5), activation='relu'),

    # S4: Average Pooling Layer (2x2)
    layers.AveragePooling2D((2, 2)),

    # C5: Flattening followed by Dense Layer
    layers.Flatten(),
    layers.Dense(120, activation='relu'),

    # F6: Fully Connected Layer
    layers.Dense(84, activation='relu'),

    # Output Layer (10 classes)
    layers.Dense(10, activation='softmax')
])

# --- 3. Compile and Train ---
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

print("Starting training...")
model.fit(X_train, y_train, epochs=10, batch_size=64, validation_split=0.1)

# --- 4. Evaluate ---
test_loss, test_acc = model.evaluate(X_test, y_test)
print(f"\nLeNet-5 Test Accuracy: {test_acc:.4f}")
```
